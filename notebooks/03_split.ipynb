# Preparacion de los datos finales

## Imports

```python
import pandas as pd
import numpy as np
import os

from sklearn.model_selection import train_test_split

from dotenv import load_dotenv

# Cargar las del archivo .env (si existe)
load_dotenv()
```

## Carga del dataset

```python
file_processed = os.path.join('..','data', 'processed',os.getenv('FILE_PROCESSED'))
df = pd.read_csv(file_processed)
```

## separar dataset

```python
# separa conjunto de datos para realizar el modelado y despues para utilizarlos en una simulacion real y evitar el data leakage
df_train_test, df_evaluacion = train_test_split(
    df,
    test_size=0.20,
    random_state=42
)
```

## Guardar datasets

```python
df_train_test.to_csv(os.path.join('..','data', 'clean',os.getenv('FILE_TRAINTEST')), index=False)

df_evaluacion.to_csv(os.path.join('..','data', 'clean',os.getenv('FILE_EVALUACION')), index=False)
```
