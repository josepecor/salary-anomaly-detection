# Procesado de Datos

## Imports

```python
import pandas as pd
import numpy as np
import os
import sys

from dotenv import load_dotenv

# Custom modules
# Obtener el directorio del notebook (notebooks/)
notebook_dir = os.getcwd()
# Subir un nivel para llegar a la raíz del proyecto
project_root = os.path.abspath(os.path.join(notebook_dir, '..'))
# Agregar al path si no está (solo afecta a la instancia actual)
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.data_management import get_categorical_number_columns, delete_columns

# Cargar las del archivo .env (si existe)
load_dotenv()
```

## Cargar Dataset

```python
#se utiliza el metodo os.path.join para construir la ruta al archivo para que sea compatible con diferentes sistemas operativos
file_raw = os.path.join('..','data', 'raw',os.getenv('FILE_RAW'))
df = pd.read_csv(file_raw)
```

```python
# Convertir columnas numéricas categóricas a tipo 'object'
categoricas_numericas = get_categorical_number_columns()

for col in categoricas_numericas:
    if col in df.columns:
        df[col] = df[col].astype("object")
```

## Eliminacion de variables

```python
# Eliminar columnas irrelevantes, funcion definida en src/data_management.py para mantener el notebook limpio y reusable
cols_eliminar = delete_columns()

print("\nShape antes de eliminar columnas irrelevantes:", df.shape)

df = df.drop(columns=cols_eliminar, errors="ignore")

print("\nShape tras eliminar columnas irrelevantes:", df.shape)
print("\nColumnas actuales:")
print(df.columns.tolist())
```

## Transformaciones y prcesado

Al tratarse de un dataset donde no se han encontrado valores nulos ni datos inconsistentes no se aplica ninguna transformacion ni procesado de los datos adicional a la eliminacion de las variables irrelevantes para el modelo.

## Guardar dataset

```python
output_path = os.path.join('..','data', 'processed', 'IBM_HR_Analytics_processed.csv')
df.to_csv(output_path, index=False)

print(f"\nDataset procesado guardado en: {output_path}")
```
